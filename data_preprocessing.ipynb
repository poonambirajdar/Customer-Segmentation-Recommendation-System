```python
import pandas as pd
```

```python
customers = pd.read_csv(r"D:\Customer segmentation project\Data/olist_customers_dataset.csv")
orders = pd.read_csv(r"D:\Customer segmentation project\Data/olist_orders_dataset.csv")
items = pd.read_csv(r"D:\Customer segmentation project\Data/olist_order_items_dataset.csv")
products = pd.read_csv(r"D:\Customer segmentation project\Data/olist_products_dataset.csv")
payments = pd.read_csv(r"D:\Customer segmentation project\Data/olist_order_payments_dataset.csv")
```

```python
df = orders.merge(customers,on="customer_id")
df = df.merge(items,on="order_id")
df = df.merge(products,on="product_id")
df = df.merge(payments,on="order_id")
```

```python
df["total_order_value"] = df["price"] + df["freight_value"]
```

```python
df.to_csv(r"D:\Customer segmentation project\processed_data/merged_dataset.csv",index=False)
```
